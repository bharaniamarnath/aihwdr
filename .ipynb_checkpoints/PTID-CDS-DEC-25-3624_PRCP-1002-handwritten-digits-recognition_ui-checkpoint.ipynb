### Data Analaysis

#### Import Packages

```python
# Import essential packages
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configurations
import warnings
warnings.filterwarnings('ignore')
```

#### Import Data

```python
# Import MNIST dataset from tensorflow keras
from tensorflow.keras.datasets import mnist

# Make Dataframe
mnist_df = pd.DataFrame(mnist.load_data())

# Make train and test data
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

#### Explore Data

```python
# Display the first 5 rows of the dataframe
print(f"\nFirst 5 rows:\n{mnist_df.head()}")

# Basic information about the dataset
print(f"\nBasic info:\n{mnist_df.info()}")

# Check for missing values
print(f"\nCheck missing values count:\n{mnist_df.isnull().sum().sum()}")

# Normalize dataset image pixel values
x_train = x_train.astype('float32') / 255
x_test = x_test.astype('float32') / 255

# Check the shape of the datasets
print("Dataset shape:\n")
print(f"Training data: {x_train.shape}")
print(f"Training labels:\n: {y_train.shape}")
print(f"Test data: {x_test.shape}")
print(f"Test labels: {y_test.shape}")
```

#### Display sample images

```python
# Display 10 sample images function
def plot_samples(images, labels, n=10):
    plt.figure(figsize=(15, 5))
    for i in range(n):
        plt.subplot(2, 5, i + 1)
        plt.imshow(images[i], cmap='gray')
        plt.title(f'Label: {labels[i]}')
        plt.axis('off')
    plt.tight_layout()
    plt.show()

# Call display sample images function
plot_samples(x_train, y_train, n=10)
```

#### Digits distribution

```python
# Training data distribution of each digit plot
plt.figure(figsize=(10, 6))
sns.countplot(x=y_train, palette='tab20')
plt.title('MNIST training data digits distribution')
plt.xlabel('Digit')
plt.ylabel('Frequency')
plt.xticks(ticks=np.arange(10), labels=np.arange(10))
plt.show()
```

#### Mean image

```python
# Training data mean image
mean_image = np.mean(x_train, axis=0)

# Display mean image
plt.figure(figsize=(5, 5))
plt.imshow(mean_image, cmap='gray')
plt.title('MNIST training data mean image')
plt.axis('off')
plt.show()
```

#### Display random images

```python
# Display training data random images with labels
indices = np.random.choice(len(x_train), 5, replace=False)
plt.figure(figsize=(15, 5))
for i, idx in enumerate(indices):
    plt.subplot(1, 5, i + 1)
    plt.imshow(x_train[idx], cmap='gray')
    plt.title(f'Label: {y_train[idx]}')
    plt.axis('off')
plt.show()
```

#### Correlation

```python
# Correlation from 10k training data images
sample_x = x_train.reshape(x_train.shape[0], 28 * 28)
sample_df = pd.DataFrame(sample_x[:10000])

# Correlation matrix
correlation_matrix = sample_df.corr()

# CM plot
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, cmap='coolwarm', square=True, cbar_kws={"shrink": .8})
plt.title('MNIST sample correlation matrix')
plt.show()
```

### Build models

```python
# Import necessary packages for models
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from PIL import Image
```

```python
# Reshape train and test data
X_train = x_train.reshape(x_train.shape[0], -1) / 255
X_test = x_test.reshape(x_test.shape[0], -1) / 255
```

```python
# Split train data for validation
X_train, X_test, y_train, y_test = train_test_split(X_train, y_train, test_size=0.1, random_state=42)
```

```python
# Scale the data for SVM and Random Forest
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
X_test_scaled = scaler.transform(X_test)
```

#### SVM Model

```python
svm_model = SVC(kernel='linear', random_state=42)
svm_model.fit(X_train_scaled, y_train)
svm_test_pred = svm_model.predict(X_test_scaled)
```

#### Random Forest Model

```python
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(X_train, y_train)
rf_test_pred = rf_model.predict(X_test)
```

#### Keras Sequential Model

```python
# Reshape train data to keras compatible format
X_train_reshaped = X_train.reshape(-1, 28, 28)
X_test_reshaped = X_test.reshape(-1, 28, 28)

keras_model = Sequential()
keras_model.add(Flatten(input_shape=(28, 28)))
keras_model.add(Dense(128, activation='relu'))
keras_model.add(Dense(10, activation='softmax'))

# Compile and fit sequential model
keras_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
keras_model.fit(X_train_reshaped, y_train, epochs=5, validation_data=(X_test_reshaped, y_test))
```

#### Prediction

```python
# Predictions and evaluations
rf_test_pred = rf_model.predict(X_test)
keras_test_pred = np.argmax(keras_model.predict(X_test.reshape(-1, 28, 28)), axis=1)
```

#### Model accuracy

```python
# Accuracy report
acc_svm = accuracy_score(y_test, svm_test_pred)
acc_rf = accuracy_score(y_test, rf_test_pred)
acc_keras = accuracy_score(y_test, keras_test_pred)

print("\nSVM Validation Accuracy:", acc_svm)
print("\nRandom Forest Validation Accuracy:", acc_rf)
print("\nKeras Model Validation Accuracy:", acc_keras)

# Calculate accuracy for each model
accuracies = {
    'SVM': acc_svm,
    'Random Forest': acc_rf,
    'Keras': acc_keras
}
```

### Compare models

```python
# Display accuracies for reference
for model, accuracy in accuracies.items():
    print(f"{model} Validation Accuracy: {accuracy:.4f}")

# Classification report
print("\nSVM Classification Report:\n", classification_report(y_test, svm_test_pred))
print("\nRandom Forest Classification Report:\n", classification_report(y_test, rf_test_pred))
print("\nKeras Model Classification Report:\n", classification_report(y_test, keras_test_pred))

# Comparison of models
models = list(accuracies.keys())
plt.bar(models, accuracies.values())
plt.ylabel('Accuracy')
plt.title('Model Accuracy Comparison')
plt.ylim([0, 1])
plt.show()
```

#### Best classifier

```python
# Find best classifier
best_classifier = max(accuracies, key=accuracies.get)
best_accuracy = accuracies[best_classifier]

print(f"The best classifier is: {best_classifier} with an accuracy of {best_accuracy:.4f}")
```

### Predict Image

#### Random Forest Classifier

```python
# Function to load and predict handwritten number image
def predict_handwritten_digit(image_path):
    # Load image and convert to grayscale
    img = Image.open(image_path).convert('L')
    # Resize to 28x28
    img_resized = img.resize((28, 28), Image.ANTIALIAS)
    
    # Convert resized image to numpy array and data type to float
    img_array = np.array(img_resized, dtype=np.float32)

    # Invert image colors
    img_array = 255 - img_array

    # Normalize and flatten
    img_array = img_array / 255.0
    # Flatten to (1, 784)
    img_flattened = img_array.flatten().reshape(1, -1)

    # Make prediction
    prediction = rf_model.predict(img_flattened)
    return prediction[0]

# Set predict image
# Predict image path
image_path = 'datasets/images/handwritten_numbers/9f.png'
# Call predict image function
predicted_digit = predict_handwritten_digit(image_path)
print(f"The predicted digit is: {predicted_digit}")

# Load and display predict image
# Convert to grayscale
img = Image.open(image_path).convert('L')
# Resize to 28x28
img_resized = img.resize((28, 28))

# Convert resized image to numpy array
img_array = np.array(img_resized)

# Display image
plt.imshow(img_array, cmap='gray')
plt.axis('off')  # Turn off axis labels
plt.show()

```
